```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

warnings.filterwarnings("ignore")

# ==============================
# Paths
# ==============================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Agricultural Rainfall Suitability Prediction System"

DATA_PATH = os.path.join(BASE_DIR, "DailyRainfallDataRaigadMaharashtra.csv")

os.makedirs(BASE_DIR, exist_ok=True)

# ==============================
# Load Dataset
# ==============================

df = pd.read_csv(DATA_PATH)

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")

df = df.dropna(subset=["Date"])

df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
df["DayOfYear"] = df["Date"].dt.dayofyear

rain_col = "Daily Rainfall (m.m)"
prog_col = "Progressive (m.m)"

df[rain_col] = df[rain_col].fillna(0)
df[prog_col] = df[prog_col].fillna(method="ffill").fillna(0)

# ==============================
# Feature Engineering
# ==============================

df["Rainfall_3Day_MA"] = df[rain_col].rolling(window=3, min_periods=1).mean()
df["Rainfall_7Day_MA"] = df[rain_col].rolling(window=7, min_periods=1).mean()
df["Rainfall_15Day_MA"] = df[rain_col].rolling(window=15, min_periods=1).mean()
df["Rainfall_Lag_1"] = df[rain_col].shift(1).fillna(0)
df["Rainfall_Lag_2"] = df[rain_col].shift(2).fillna(0)
df["Rainfall_Lag_3"] = df[rain_col].shift(3).fillna(0)

# ==============================
# Crop Suitability Label Creation
# ==============================

def crop_suitability(row):
    rainfall = row[rain_col]
    ma7 = row["Rainfall_7Day_MA"]

    if rainfall >= 80 or ma7 >= 60:
        return "Rice Suitable"
    elif rainfall >= 40 or ma7 >= 35:
        return "Sugarcane Suitable"
    elif rainfall >= 15 or ma7 >= 15:
        return "Pulses Suitable"
    else:
        return "Low Water Crops Suitable"

df["Crop_Suitability"] = df.apply(crop_suitability, axis=1)

# ==============================
# Features and Target
# ==============================

features = [
    rain_col,
    prog_col,
    "Day",
    "Month",
    "Year",
    "DayOfYear",
    "Rainfall_3Day_MA",
    "Rainfall_7Day_MA",
    "Rainfall_15Day_MA",
    "Rainfall_Lag_1",
    "Rainfall_Lag_2",
    "Rainfall_Lag_3"
]

X = df[features]
y = df["Crop_Suitability"]

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# ==============================
# Random Forest Model
# ==============================

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=8
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

# ==============================
# Neural Network Model
# ==============================

num_classes = len(label_encoder.classes_)

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

nn_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

nn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = nn_model.fit(
    X_train,
    y_train_cat,
    validation_data=(X_test, y_test_cat),
    epochs=100,
    batch_size=8,
    verbose=1
)

nn_loss, nn_accuracy = nn_model.evaluate(X_test, y_test_cat, verbose=0)

nn_pred_prob = nn_model.predict(X_scaled)
nn_pred = np.argmax(nn_pred_prob, axis=1)

df["Predicted_Crop_Suitability"] = label_encoder.inverse_transform(nn_pred)

# ==============================
# Save Models
# ==============================

nn_model.save(os.path.join(BASE_DIR, "rainfall_suitability_model.h5"))

with open(os.path.join(BASE_DIR, "rainfall_random_forest_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, "rainfall_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, "rainfall_label_encoder.pkl"), "wb") as f:
    pickle.dump(label_encoder, f)

# ==============================
# Save CSV Outputs
# ==============================

df.to_csv(os.path.join(BASE_DIR, "rainfall_suitability_predictions.csv"), index=False)

metrics_df = pd.DataFrame({
    "Model": ["Random Forest", "Neural Network"],
    "Accuracy": [rf_accuracy, nn_accuracy]
})

metrics_df.to_csv(os.path.join(BASE_DIR, "rainfall_suitability_accuracy.csv"), index=False)

feature_importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

feature_importance_df.to_csv(
    os.path.join(BASE_DIR, "rainfall_feature_importance.csv"),
    index=False
)

# ==============================
# Save JSON Result
# ==============================

results = {
    "project_name": "Agricultural Rainfall Suitability Prediction System",
    "dataset": DATA_PATH,
    "total_records": int(len(df)),
    "features_used": features,
    "target_column": "Crop_Suitability",
    "classes": label_encoder.classes_.tolist(),
    "random_forest_accuracy": float(rf_accuracy),
    "neural_network_accuracy": float(nn_accuracy),
    "output_files": [
        "rainfall_suitability_model.h5",
        "rainfall_random_forest_model.pkl",
        "rainfall_scaler.pkl",
        "rainfall_label_encoder.pkl",
        "rainfall_suitability_predictions.csv",
        "rainfall_suitability_accuracy.csv",
        "rainfall_feature_importance.csv"
    ]
}

with open(os.path.join(BASE_DIR, "rainfall_suitability_results.json"), "w") as f:
    json.dump(results, f, indent=4)

# ==============================
# Save YAML Config
# ==============================

config = {
    "project": "Agricultural Rainfall Suitability Prediction System",
    "author": "Sagnik Patra",
    "dataset_path": DATA_PATH,
    "base_directory": BASE_DIR,
    "models": {
        "random_forest": {
            "n_estimators": 200,
            "max_depth": 8,
            "random_state": 42
        },
        "neural_network": {
            "epochs": 100,
            "batch_size": 8,
            "optimizer": "adam",
            "loss": "categorical_crossentropy"
        }
    },
    "features": features,
    "target": "Crop_Suitability",
    "classes": label_encoder.classes_.tolist()
}

with open(os.path.join(BASE_DIR, "rainfall_suitability_config.yaml"), "w") as f:
    yaml.dump(config, f, sort_keys=False)

# ==============================
# Graphs
# ==============================

plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df[rain_col])
plt.title("Daily Rainfall Trend")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "rainfall_trend_graph.png"))
plt.close()

plt.figure(figsize=(8, 5))
plt.bar(metrics_df["Model"], metrics_df["Accuracy"])
plt.title("Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "rainfall_accuracy_graph.png"))
plt.close()

plt.figure(figsize=(10, 6))
plt.barh(feature_importance_df["Feature"], feature_importance_df["Importance"])
plt.title("Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "rainfall_feature_importance_graph.png"))
plt.close()

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(7, 6))
plt.imshow(cm)
plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), label_encoder.classes_, rotation=45, ha="right")
plt.yticks(range(num_classes), label_encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "rainfall_confusion_matrix.png"))
plt.close()

# ==============================
# Print Summary
# ==============================

print("\nAgricultural Rainfall Suitability Prediction System Completed Successfully")
print("Random Forest Accuracy:", rf_accuracy)
print("Neural Network Accuracy:", nn_accuracy)
print("\nFiles saved in:")
print(BASE_DIR)
```
