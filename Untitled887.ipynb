```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

warnings.filterwarnings("ignore")

# ===============================
# Project Paths
# ===============================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Agricultural Rainfall Suitability Prediction System"
DATA_PATH = os.path.join(BASE_DIR, "DailyRainfallDataRaigadMaharashtra.csv")

os.makedirs(BASE_DIR, exist_ok=True)

PREFIX = "ais_"

# ===============================
# Load Dataset
# ===============================

df = pd.read_csv(DATA_PATH)

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")
df = df.dropna(subset=["Date"])

RAIN_COL = "Daily Rainfall (m.m)"
PROG_COL = "Progressive (m.m)"

df[RAIN_COL] = df[RAIN_COL].fillna(0)
df[PROG_COL] = df[PROG_COL].fillna(method="ffill").fillna(0)

# ===============================
# Feature Engineering
# ===============================

df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
df["DayOfYear"] = df["Date"].dt.dayofyear
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)

df["Rainfall_Lag_1"] = df[RAIN_COL].shift(1).fillna(0)
df["Rainfall_Lag_2"] = df[RAIN_COL].shift(2).fillna(0)
df["Rainfall_Lag_3"] = df[RAIN_COL].shift(3).fillna(0)

df["Rainfall_3Day_MA"] = df[RAIN_COL].rolling(3, min_periods=1).mean()
df["Rainfall_7Day_MA"] = df[RAIN_COL].rolling(7, min_periods=1).mean()
df["Rainfall_15Day_MA"] = df[RAIN_COL].rolling(15, min_periods=1).mean()
df["Rainfall_30Day_MA"] = df[RAIN_COL].rolling(30, min_periods=1).mean()

df["Weekly_Rainfall"] = df[RAIN_COL].rolling(7, min_periods=1).sum()
df["Monthly_Rainfall"] = df[RAIN_COL].rolling(30, min_periods=1).sum()

# ===============================
# Target Creation
# ===============================

def crop_suitability(row):
    rain = row[RAIN_COL]
    weekly = row["Weekly_Rainfall"]
    monthly = row["Monthly_Rainfall"]

    if rain >= 80 or weekly >= 350 or monthly >= 900:
        return "Rice Suitable"
    elif rain >= 45 or weekly >= 220 or monthly >= 600:
        return "Sugarcane Suitable"
    elif rain >= 15 or weekly >= 80 or monthly >= 250:
        return "Pulses Suitable"
    else:
        return "Low Water Crops Suitable"

df["Crop_Suitability"] = df.apply(crop_suitability, axis=1)

# ===============================
# Crop Planning
# ===============================

def crop_planning(row):
    crop = row["Crop_Suitability"]

    if crop == "Rice Suitable":
        return "Plan rice cultivation; ensure drainage during extreme rainfall"
    elif crop == "Sugarcane Suitable":
        return "Plan sugarcane cultivation with controlled irrigation"
    elif crop == "Pulses Suitable":
        return "Plan pulses or short-duration crops"
    else:
        return "Plan millet, pulses, or low-water crops"

df["Crop_Planning_Recommendation"] = df.apply(crop_planning, axis=1)

# ===============================
# Features
# ===============================

features = [
    RAIN_COL,
    PROG_COL,
    "Day",
    "Month",
    "Year",
    "DayOfYear",
    "Week",
    "Rainfall_Lag_1",
    "Rainfall_Lag_2",
    "Rainfall_Lag_3",
    "Rainfall_3Day_MA",
    "Rainfall_7Day_MA",
    "Rainfall_15Day_MA",
    "Rainfall_30Day_MA",
    "Weekly_Rainfall",
    "Monthly_Rainfall"
]

X = df[features]
y = df["Crop_Suitability"]

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train_full, X_test_full, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# ===============================
# AIS Feature Selection
# ===============================

def evaluate_antibody(mask):
    selected = np.where(mask == 1)[0]

    if len(selected) == 0:
        return 0

    X_train = X_train_full[:, selected]
    X_test = X_test_full[:, selected]

    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=8,
        random_state=42
    )

    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    acc = accuracy_score(y_test, pred)

    feature_penalty = len(selected) / len(mask) * 0.02

    fitness = acc - feature_penalty

    return fitness


def mutate_antibody(mask, mutation_rate=0.15):
    new_mask = mask.copy()

    for i in range(len(new_mask)):
        if np.random.rand() < mutation_rate:
            new_mask[i] = 1 - new_mask[i]

    if np.sum(new_mask) == 0:
        new_mask[np.random.randint(0, len(new_mask))] = 1

    return new_mask


def ais_feature_selection(
    num_features,
    population_size=20,
    generations=30,
    clone_factor=5,
    mutation_rate=0.15
):
    population = np.random.randint(0, 2, size=(population_size, num_features))

    for i in range(population_size):
        if np.sum(population[i]) == 0:
            population[i][np.random.randint(0, num_features)] = 1

    best_mask = None
    best_fitness = -np.inf
    fitness_history = []

    for generation in range(generations):
        fitness_scores = np.array([evaluate_antibody(mask) for mask in population])

        best_idx = np.argmax(fitness_scores)

        if fitness_scores[best_idx] > best_fitness:
            best_fitness = fitness_scores[best_idx]
            best_mask = population[best_idx].copy()

        sorted_indices = np.argsort(fitness_scores)[::-1]
        top_population = population[sorted_indices[:population_size // 2]]

        clones = []

        for antibody in top_population:
            for _ in range(clone_factor):
                clone = mutate_antibody(antibody, mutation_rate)
                clones.append(clone)

        clones = np.array(clones)

        clone_fitness = np.array([evaluate_antibody(mask) for mask in clones])

        combined_population = np.vstack((population, clones))
        combined_fitness = np.concatenate((fitness_scores, clone_fitness))

        selected_indices = np.argsort(combined_fitness)[::-1][:population_size]
        population = combined_population[selected_indices]

        random_new = np.random.randint(0, 2, size=(2, num_features))

        for i in range(len(random_new)):
            if np.sum(random_new[i]) == 0:
                random_new[i][np.random.randint(0, num_features)] = 1

        population[-2:] = random_new

        fitness_history.append(best_fitness)

        print(f"Generation {generation + 1}/{generations} | Best Fitness: {best_fitness:.4f}")

    return best_mask, best_fitness, fitness_history


best_mask, best_fitness, fitness_history = ais_feature_selection(
    num_features=len(features),
    population_size=20,
    generations=30,
    clone_factor=5,
    mutation_rate=0.15
)

selected_feature_indices = np.where(best_mask == 1)[0]
selected_features = [features[i] for i in selected_feature_indices]

X_selected = X_scaled[:, selected_feature_indices]

X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# ===============================
# AIS Optimized Random Forest
# ===============================

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

# ===============================
# AIS Optimized Neural Network
# ===============================

num_classes = len(label_encoder.classes_)

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

nn_model = Sequential([
    Dense(128, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(num_classes, activation="softmax")
])

nn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = nn_model.fit(
    X_train,
    y_train_cat,
    validation_data=(X_test, y_test_cat),
    epochs=100,
    batch_size=8,
    verbose=1
)

nn_loss, nn_accuracy = nn_model.evaluate(X_test, y_test_cat, verbose=0)

# ===============================
# Final Prediction
# ===============================

final_pred_encoded = rf_model.predict(X_selected)
df["Predicted_Crop_Suitability"] = label_encoder.inverse_transform(final_pred_encoded)

df["Prediction_Status"] = np.where(
    df["Crop_Suitability"] == df["Predicted_Crop_Suitability"],
    "Correct",
    "Incorrect"
)

# ===============================
# Save Models
# ===============================

nn_model.save(os.path.join(BASE_DIR, PREFIX + "crop_suitability_model.h5"))

with open(os.path.join(BASE_DIR, PREFIX + "crop_random_forest_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, PREFIX + "crop_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, PREFIX + "crop_label_encoder.pkl"), "wb") as f:
    pickle.dump(label_encoder, f)

# ===============================
# Save CSV Files
# ===============================

df.to_csv(os.path.join(BASE_DIR, PREFIX + "crop_prediction.csv"), index=False)

result_df = pd.DataFrame({
    "Model": ["AIS-Random Forest", "AIS-Neural Network"],
    "Accuracy": [rf_accuracy, nn_accuracy]
})

result_df.to_csv(os.path.join(BASE_DIR, PREFIX + "crop_result.csv"), index=False)

planning_df = df[
    [
        "District",
        "Date",
        RAIN_COL,
        PROG_COL,
        "Weekly_Rainfall",
        "Monthly_Rainfall",
        "Crop_Suitability",
        "Predicted_Crop_Suitability",
        "Crop_Planning_Recommendation"
    ]
]

planning_df.to_csv(
    os.path.join(BASE_DIR, PREFIX + "crop_planning_rainfall_analytics.csv"),
    index=False
)

selected_features_df = pd.DataFrame({
    "Selected_Feature": selected_features
})

selected_features_df.to_csv(
    os.path.join(BASE_DIR, PREFIX + "selected_features.csv"),
    index=False
)

fitness_df = pd.DataFrame({
    "Generation": list(range(1, len(fitness_history) + 1)),
    "Best_Fitness": fitness_history
})

fitness_df.to_csv(
    os.path.join(BASE_DIR, PREFIX + "optimization_fitness.csv"),
    index=False
)

report = classification_report(
    y_test,
    rf_pred,
    target_names=label_encoder.classes_,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()
report_df.to_csv(os.path.join(BASE_DIR, PREFIX + "classification_report.csv"))

# ===============================
# JSON Results
# ===============================

results_json = {
    "project_name": "Agricultural Rainfall Suitability Prediction System using AIS",
    "author": "Sagnik Patra",
    "dataset": DATA_PATH,
    "algorithm": "Artificial Immune System Feature Selection + Random Forest + Neural Network",
    "total_records": int(len(df)),
    "total_features": len(features),
    "selected_features": selected_features,
    "best_ais_fitness": float(best_fitness),
    "random_forest_accuracy": float(rf_accuracy),
    "neural_network_accuracy": float(nn_accuracy),
    "classes": label_encoder.classes_.tolist()
}

with open(os.path.join(BASE_DIR, PREFIX + "results.json"), "w") as f:
    json.dump(results_json, f, indent=4)

# ===============================
# YAML Config
# ===============================

config_yaml = {
    "project": "Agricultural Rainfall Suitability Prediction System using AIS",
    "author": "Sagnik Patra",
    "dataset_path": DATA_PATH,
    "base_directory": BASE_DIR,
    "prefix": PREFIX,
    "optimization": {
        "algorithm": "Artificial Immune System",
        "population_size": 20,
        "generations": 30,
        "clone_factor": 5,
        "mutation_rate": 0.15
    },
    "models": {
        "random_forest": {
            "n_estimators": 300,
            "max_depth": 10,
            "random_state": 42
        },
        "neural_network": {
            "epochs": 100,
            "batch_size": 8,
            "optimizer": "adam"
        }
    },
    "selected_features": selected_features,
    "classes": label_encoder.classes_.tolist()
}

with open(os.path.join(BASE_DIR, PREFIX + "config.yaml"), "w") as f:
    yaml.dump(config_yaml, f, sort_keys=False)

# ===============================
# Graphs
# ===============================

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("AIS Model Accuracy Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "accuracy_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(result_df["Model"], result_df["Accuracy"], marker="o")
plt.title("AIS Model Comparison Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "comparison_graph.png"))
plt.show()

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(8, 6))
plt.imshow(cm)
plt.title("AIS Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), label_encoder.classes_, rotation=45, ha="right")
plt.yticks(range(num_classes), label_encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "heatmap_confusion_matrix.png"))
plt.show()

crop_counts = df["Predicted_Crop_Suitability"].value_counts()

plt.figure(figsize=(9, 5))
plt.bar(crop_counts.index, crop_counts.values)
plt.title("AIS Result Graph - Predicted Crop Suitability")
plt.xlabel("Crop Suitability Class")
plt.ylabel("Number of Days")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "result_graph.png"))
plt.show()

plt.figure(figsize=(12, 5))
plt.plot(df["Date"], df[RAIN_COL])
plt.title("AIS Prediction Graph - Daily Rainfall Pattern")
plt.xlabel("Date")
plt.ylabel("Rainfall mm")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "prediction_graph.png"))
plt.show()

planning_counts = df["Crop_Planning_Recommendation"].value_counts()

plt.figure(figsize=(10, 6))
plt.barh(planning_counts.index, planning_counts.values)
plt.title("AIS Crop Planning using Rainfall Analytics")
plt.xlabel("Number of Days")
plt.ylabel("Crop Planning Recommendation")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "crop_planning_analytics_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(fitness_history) + 1), fitness_history, marker="o")
plt.title("AIS Optimization Fitness Graph")
plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "optimization_fitness_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("AIS Neural Network Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "nn_training_accuracy_graph.png"))
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("AIS Neural Network Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, PREFIX + "nn_training_loss_graph.png"))
plt.show()

print("\nAIS Agricultural Rainfall Suitability Prediction Completed Successfully!")
print("Selected Features:", selected_features)
print("AIS Best Fitness:", best_fitness)
print("Random Forest Accuracy:", rf_accuracy)
print("Neural Network Accuracy:", nn_accuracy)
print("All files saved with ais_ prefix in:")
print(BASE_DIR)
```
