```python
import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

warnings.filterwarnings("ignore")

# ===============================
# Project Paths
# ===============================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Agricultural Rainfall Suitability Prediction System"
DATA_PATH = os.path.join(BASE_DIR, "DailyRainfallDataRaigadMaharashtra.csv")

os.makedirs(BASE_DIR, exist_ok=True)

# ===============================
# Load Data
# ===============================

df = pd.read_csv(DATA_PATH)

df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")
df = df.dropna(subset=["Date"])

RAIN_COL = "Daily Rainfall (m.m)"
PROG_COL = "Progressive (m.m)"

df[RAIN_COL] = df[RAIN_COL].fillna(0)
df[PROG_COL] = df[PROG_COL].fillna(method="ffill").fillna(0)

# ===============================
# Feature Engineering
# ===============================

df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
df["DayOfYear"] = df["Date"].dt.dayofyear
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)

df["Rainfall_Lag_1"] = df[RAIN_COL].shift(1).fillna(0)
df["Rainfall_Lag_2"] = df[RAIN_COL].shift(2).fillna(0)
df["Rainfall_Lag_3"] = df[RAIN_COL].shift(3).fillna(0)

df["Rainfall_3Day_MA"] = df[RAIN_COL].rolling(3, min_periods=1).mean()
df["Rainfall_7Day_MA"] = df[RAIN_COL].rolling(7, min_periods=1).mean()
df["Rainfall_15Day_MA"] = df[RAIN_COL].rolling(15, min_periods=1).mean()
df["Rainfall_30Day_MA"] = df[RAIN_COL].rolling(30, min_periods=1).mean()

df["Weekly_Rainfall"] = df[RAIN_COL].rolling(7, min_periods=1).sum()
df["Monthly_Rainfall"] = df[RAIN_COL].rolling(30, min_periods=1).sum()

# ===============================
# Crop Suitability Target
# ===============================

def crop_suitability(row):
    rain = row[RAIN_COL]
    weekly = row["Weekly_Rainfall"]
    monthly = row["Monthly_Rainfall"]

    if rain >= 80 or weekly >= 350 or monthly >= 900:
        return "Rice Suitable"
    elif rain >= 45 or weekly >= 220 or monthly >= 600:
        return "Sugarcane Suitable"
    elif rain >= 15 or weekly >= 80 or monthly >= 250:
        return "Pulses Suitable"
    else:
        return "Low Water Crops Suitable"

df["Crop_Suitability"] = df.apply(crop_suitability, axis=1)

# ===============================
# Crop Planning Analytics
# ===============================

def crop_planning(row):
    crop = row["Crop_Suitability"]

    if crop == "Rice Suitable":
        return "Plan rice cultivation; ensure drainage during extreme rainfall"
    elif crop == "Sugarcane Suitable":
        return "Plan sugarcane cultivation with controlled irrigation"
    elif crop == "Pulses Suitable":
        return "Plan pulses or short-duration crops"
    else:
        return "Plan millet, pulses, or low-water crops"

df["Crop_Planning_Recommendation"] = df.apply(crop_planning, axis=1)

# ===============================
# Features
# ===============================

features = [
    RAIN_COL,
    PROG_COL,
    "Day",
    "Month",
    "Year",
    "DayOfYear",
    "Week",
    "Rainfall_Lag_1",
    "Rainfall_Lag_2",
    "Rainfall_Lag_3",
    "Rainfall_3Day_MA",
    "Rainfall_7Day_MA",
    "Rainfall_15Day_MA",
    "Rainfall_30Day_MA",
    "Weekly_Rainfall",
    "Monthly_Rainfall"
]

X = df[features]
y = df["Crop_Suitability"]

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# ===============================
# Random Forest Model
# ===============================

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42
)

rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_pred)

# ===============================
# Gradient Boosting Model
# ===============================

gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train, y_train)
gb_pred = gb_model.predict(X_test)
gb_accuracy = accuracy_score(y_test, gb_pred)

# ===============================
# Neural Network Model
# ===============================

num_classes = len(label_encoder.classes_)

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

nn_model = Sequential([
    Dense(128, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(num_classes, activation="softmax")
])

nn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = nn_model.fit(
    X_train,
    y_train_cat,
    validation_data=(X_test, y_test_cat),
    epochs=100,
    batch_size=8,
    verbose=1
)

nn_loss, nn_accuracy = nn_model.evaluate(X_test, y_test_cat, verbose=0)

# ===============================
# Final Prediction
# ===============================

final_pred_encoded = rf_model.predict(X_scaled)
df["Predicted_Crop_Suitability"] = label_encoder.inverse_transform(final_pred_encoded)

df["Prediction_Status"] = np.where(
    df["Crop_Suitability"] == df["Predicted_Crop_Suitability"],
    "Correct",
    "Incorrect"
)

# ===============================
# Save Models
# ===============================

nn_model.save(os.path.join(BASE_DIR, "crop_suitability_model.h5"))

with open(os.path.join(BASE_DIR, "crop_random_forest_model.pkl"), "wb") as f:
    pickle.dump(rf_model, f)

with open(os.path.join(BASE_DIR, "crop_gradient_boosting_model.pkl"), "wb") as f:
    pickle.dump(gb_model, f)

with open(os.path.join(BASE_DIR, "crop_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

with open(os.path.join(BASE_DIR, "crop_label_encoder.pkl"), "wb") as f:
    pickle.dump(label_encoder, f)

# ===============================
# Save CSV Files
# ===============================

prediction_csv = os.path.join(BASE_DIR, "crop_prediction.csv")
result_csv = os.path.join(BASE_DIR, "crop_result.csv")
planning_csv = os.path.join(BASE_DIR, "crop_planning_rainfall_analytics.csv")
accuracy_csv = os.path.join(BASE_DIR, "crop_accuracy_comparison.csv")

df.to_csv(prediction_csv, index=False)

result_df = pd.DataFrame({
    "Model": ["Random Forest", "Gradient Boosting", "Neural Network"],
    "Accuracy": [rf_accuracy, gb_accuracy, nn_accuracy]
})

result_df.to_csv(result_csv, index=False)
result_df.to_csv(accuracy_csv, index=False)

crop_planning_df = df[
    [
        "District",
        "Date",
        RAIN_COL,
        PROG_COL,
        "Weekly_Rainfall",
        "Monthly_Rainfall",
        "Crop_Suitability",
        "Predicted_Crop_Suitability",
        "Crop_Planning_Recommendation"
    ]
]

crop_planning_df.to_csv(planning_csv, index=False)

# ===============================
# Classification Report CSV
# ===============================

report = classification_report(
    y_test,
    rf_pred,
    target_names=label_encoder.classes_,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()
report_df.to_csv(os.path.join(BASE_DIR, "crop_classification_report.csv"))

# ===============================
# Feature Importance CSV
# ===============================

feature_importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
}).sort_values(by="Importance", ascending=False)

feature_importance_df.to_csv(
    os.path.join(BASE_DIR, "crop_feature_importance.csv"),
    index=False
)

# ===============================
# JSON Results
# ===============================

results_json = {
    "project_name": "Agricultural Rainfall Suitability Prediction System",
    "author": "Sagnik Patra",
    "dataset": DATA_PATH,
    "total_records": int(len(df)),
    "features_used": features,
    "target": "Crop_Suitability",
    "classes": label_encoder.classes_.tolist(),
    "random_forest_accuracy": float(rf_accuracy),
    "gradient_boosting_accuracy": float(gb_accuracy),
    "neural_network_accuracy": float(nn_accuracy),
    "saved_files": [
        "crop_suitability_model.h5",
        "crop_random_forest_model.pkl",
        "crop_gradient_boosting_model.pkl",
        "crop_scaler.pkl",
        "crop_label_encoder.pkl",
        "crop_prediction.csv",
        "crop_result.csv",
        "crop_planning_rainfall_analytics.csv",
        "crop_accuracy_comparison.csv",
        "crop_classification_report.csv",
        "crop_feature_importance.csv"
    ]
}

with open(os.path.join(BASE_DIR, "crop_results.json"), "w") as f:
    json.dump(results_json, f, indent=4)

# ===============================
# YAML Config
# ===============================

config_yaml = {
    "project": "Agricultural Rainfall Suitability Prediction System",
    "author": "Sagnik Patra",
    "dataset_path": DATA_PATH,
    "base_directory": BASE_DIR,
    "models": {
        "random_forest": {
            "n_estimators": 300,
            "max_depth": 10,
            "random_state": 42
        },
        "gradient_boosting": {
            "random_state": 42
        },
        "neural_network": {
            "epochs": 100,
            "batch_size": 8,
            "optimizer": "adam",
            "loss": "categorical_crossentropy"
        }
    },
    "features": features,
    "target": "Crop_Suitability",
    "classes": label_encoder.classes_.tolist()
}

with open(os.path.join(BASE_DIR, "crop_config.yaml"), "w") as f:
    yaml.dump(config_yaml, f, sort_keys=False)

# ===============================
# Graph 1: Accuracy Graph
# ===============================

plt.figure(figsize=(8, 5))
plt.bar(result_df["Model"], result_df["Accuracy"])
plt.title("Model Accuracy Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_accuracy_graph.png"))
plt.show()

# ===============================
# Graph 2: Comparison Graph
# ===============================

plt.figure(figsize=(9, 5))
plt.plot(result_df["Model"], result_df["Accuracy"], marker="o")
plt.title("Model Comparison Graph")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_comparison_graph.png"))
plt.show()

# ===============================
# Graph 3: Heatmap
# ===============================

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(8, 6))
plt.imshow(cm)
plt.title("Crop Suitability Confusion Matrix Heatmap")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks(range(num_classes), label_encoder.classes_, rotation=45, ha="right")
plt.yticks(range(num_classes), label_encoder.classes_)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_heatmap_confusion_matrix.png"))
plt.show()

# ===============================
# Graph 4: Result Graph
# ===============================

crop_counts = df["Predicted_Crop_Suitability"].value_counts()

plt.figure(figsize=(9, 5))
plt.bar(crop_counts.index, crop_counts.values)
plt.title("Result Graph - Predicted Crop Suitability Count")
plt.xlabel("Crop Suitability Class")
plt.ylabel("Number of Days")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_result_graph.png"))
plt.show()

# ===============================
# Graph 5: Prediction Graph
# ===============================

plt.figure(figsize=(12, 5))
plt.plot(df["Date"], df[RAIN_COL], label="Daily Rainfall")
plt.title("Prediction Graph - Daily Rainfall Pattern")
plt.xlabel("Date")
plt.ylabel("Rainfall mm")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_prediction_graph.png"))
plt.show()

# ===============================
# Graph 6: Crop Planning Analytics Graph
# ===============================

planning_counts = df["Crop_Planning_Recommendation"].value_counts()

plt.figure(figsize=(10, 6))
plt.barh(planning_counts.index, planning_counts.values)
plt.title("Crop Planning using Rainfall Analytics")
plt.xlabel("Number of Days")
plt.ylabel("Crop Planning Recommendation")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_planning_analytics_graph.png"))
plt.show()

# ===============================
# Graph 7: Feature Importance Graph
# ===============================

plt.figure(figsize=(10, 7))
plt.barh(feature_importance_df["Feature"], feature_importance_df["Importance"])
plt.title("Feature Importance Graph")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_feature_importance_graph.png"))
plt.show()

# ===============================
# Graph 8: Training Accuracy Graph
# ===============================

plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Neural Network Training Accuracy Graph")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_nn_training_accuracy_graph.png"))
plt.show()

# ===============================
# Graph 9: Training Loss Graph
# ===============================

plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Neural Network Training Loss Graph")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(BASE_DIR, "crop_nn_training_loss_graph.png"))
plt.show()

print("\nProject completed successfully!")
print("All CSV, H5, PKL, JSON, YAML, and graph files saved in:")
print(BASE_DIR)
```
